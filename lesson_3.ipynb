##*A sentence as a string*

```python
sentence = "All that remains is what remains"
```

```python
isinstance(sentence,int)
```

```python
isinstance(sentence,str)
```

```python
#length
len(sentence)
```

```python
#count a particular word appears
print(sentence.count("remains"))
```

```python
#count a letter appears
print(sentence.count("a"))
```

```python
#count how many times a letter appears after a certain position
print(sentence.count("a",10))
```

```python
#find position of first occurence of a particular word
print(sentence.find("remains"))
```

```python
print(sentence.find("a"))
```

```python
#reverse position -- from right to left
sentence.rfind("a")
```

```python
#find a word that is not present
sentence.find("saw")
```

```python
#to remove a word from the end of a string --from reverse
sentence.rstrip("ains")
```

```python
sentence.strip("All t")
```
