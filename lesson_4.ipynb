```python
#Taking DNA seq as input
dna = input("Enter a DNA sequence: ")
```

```python
#calculating length of input DNA sequence
length = len(dna)
length
```

```python
#converting to uppercase
dna = dna.upper()
dna
```

```python
#counting A, T, G, C
a = dna.count("A")
t = dna.count("T")
g = dna.count("G")
c = dna.count("C")
n = dna.count("N")
```

```python
print("Length= ", length, "\n",
      "N",n, "\n",
      "G",g, "\n",
      "C",c, "\n",
      "A",a, "\n",
      "T",t ,"\n",
      )
```

```python
#calculating G-C % in seq
gc = ((g + c)/length)*100
gc
```

```python
print("GC content = ", gc,"%")
```

```python
#converting to RNA
rna = dna.replace("T", "U")
rna
```

```python
print("RNA Sequence: ", rna)
```

```python
#find first occurence of AUG -substring in a string
start = rna.find("AUG")
start+1
```

```python
#find first occurence of AUG from the reverse or 3' end
start2 = rna.rfind("AUG")
start2+1
```

```python
#starts searching for pattern from x position onwards
start3 = rna.find("AUG", 5)
start3+1
```

```python
#calculate primer melting temp
tm = (4*(g+c)+2*(a+t))
print("The melting temperature of the oligo is ", tm)
```
