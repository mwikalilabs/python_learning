```python
dna2 = "AATTGGCC"
for i in dna2:
  #check for purine
  if i == "A" or i == "G":
    print(i,"purine")
  else :
    print(i,"pyrimidine")
```

```python
#counting in loop
purine=0;pyrimidine=0
dna = "GAATTCAAAA"
#loop
for i in dna:
  if i == "G" or i == "A":
    print(i,"purine")
    #to count
    purine = purine + 1
  else:
    print(i,"pyrimidine")
    #to count
    pyrimidine = pyrimidine + 1
print("Total purines= ",purine,"\n","Total pyrimidines= ",pyrimidine)
```
