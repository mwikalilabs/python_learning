Generating the reverse ***complement***

```python
seq = input("Enter a DNA sequence: ")
```

```python
seq_len = (len(seq))
print(f"Sequence:{seq} , Length is : {seq_len}")
```

```python
#converting DNA seq to RNA seq
rna = seq.replace("T","U")
print(rna)
```

```python
seq.upper()
```

```python
seq.lower()
```

```python
#complement - substitute A/T, T/A, G/C and C/G
print(seq)
complement1 = seq.replace("A","T")
print(complement1)
complement2 = complement1.replace("T","A")
print(complement2)
complement3 = complement2.replace("G","C")
print(complement3)
complement4 = complement3.replace("C","G")
print(complement4)
```

```python
print("Original Sequence: ", seq)
complement = seq.replace("A","t").replace("C","g").replace("T","a").replace("G","c")
print(complement)
print(complement.upper())
```

reverse ***compliment***

```python
reverse_complement = complement[::-1]
print("Original Seq: ",seq)
print("Reverse compliment of the sequence is: ",reverse_complement.upper())
```
