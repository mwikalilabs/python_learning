DNA Sequence Comparison - Calculating the Hamming Distance

```python
#seq alignments -#Hamming distance -HD
#The number of positions where two strings of same length differ from each other
```

```python
dna1 = input("Enter the first DNA sequence: ")
dna2 = input("Enter the second DNA sequence: ")
```

```python
#how to access individual nucleotides in a DNA seq
#first
print(dna1[0])
print(dna2[0])
#last
print(dna1[-1])
print(dna2[-1])
```

```python
#calculating HD
#checking for equal length
print(dna1)
print(dna2)
if len(dna1)==len(dna2):
  count=0
  for i in range(len(dna1)):
    if dna1[i]!=dna2[i]:
      count+=1
      print(i,dna1[i],dna2[i])
  print("HD=",count)
else:
  print("Sequences are not of equal length/DNA length mismatch")

```
